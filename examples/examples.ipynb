```python
from PIL import Image

from bbox_vis import draw_bbox, draw_multiple_bboxes
```

# Draw single box

```python
# Create a blank image, the library never draw image inplace
image = Image.new("RGB", (400, 400), (255, 255, 255))
xyxy_bboxes = [[-10, 5, 150, 150], [200, 200, 300, 300], [350, 350, 400, 400]]
labels = ["Box 1", "Box 2", "Box 3 with a very long label"]
colors = [(255, 0, 0), (0, 255, 0), (0, 100, 100)]
```

```python
# The library will always try to draw label within the image, no matter the bbox coordinates
drawn_image = draw_bbox(image, xyxy_bboxes[0], color=colors[0], label=labels[0])
drawn_image = draw_bbox(
    drawn_image, xyxy_bboxes[-1], color=colors[-1], label=labels[-1]
)
drawn_image
```

```python
# optionally, you can specify the label position to be within the bbox
draw_bbox(image, xyxy_bboxes[1], color=colors[1], label=labels[1], label_top=False)
```
